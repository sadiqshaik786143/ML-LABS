```python
def bag_of_words_vector(tokens, vocabulary):  # Define a function that creates word-count features.(AP24110011603)
    token_array = np.array(tokens)  # Convert the token list into a NumPy array for vectorized comparison.
    vector = np.array([np.sum(token_array == word) for word in vocabulary])  # Count each vocabulary word.
    return vector  # Return the bag-of-words vector.
vec = bag_of_words_vector(tokens, vocabulary)  # Create the vector for the demonstration sentence.
print("Vocabulary:", vocabulary)  # Display the vocabulary order.
print("Bag-of-words vector:", vec)  # Display the corresponding counts.
```

```python
sentence_a = "machine learning uses numpy arrays"  # Store sentence A.
sentence_b = "deep learning also uses numpy arrays heavily"  # Store sentence B.
tokens_a = sentence_a.lower().split()  # Tokenize sentence A.(AP24110011603)
tokens_b = sentence_b.lower().split()  # Tokenize sentence B.
combined_vocab = np.unique(np.array(tokens_a + tokens_b))  # Build one shared vocabulary.
vec_a = bag_of_words_vector(tokens_a, combined_vocab)  # Convert sentence A to a count vector.
vec_b = bag_of_words_vector(tokens_b, combined_vocab)  # Convert sentence B to a count vector.
print("Vector A:", vec_a)  # Display vector A.
print("Vector B:", vec_b)  # Display vector B.
cosine_sim = np.dot(vec_a, vec_b) / (np.linalg.norm(vec_a) * np.linalg.norm(vec_b))  # Calculate cosine similarity.
print("Cosine similarity:", cosine_sim)  # Display the similarity score.
```

Explanation(AP24110011603)

 Cosine similarity is close to 1 when vectors point in similar directions and close to 0 when they are nearly orthogonal. Exactly 1 means the two nonzero vectors point in exactly the same direction.


```python
sentence_c = "machine learning uses data"  # Create a sentence related to machine learning.
sentence_d = "data science uses machine learning"  # Create a similar-topic sentence.
sentence_e = "the football match was exciting"  # Create a very different-topic sentence.
def cosine_for_pair(text1, text2):  # Define a reusable cosine-similarity calculation.
    t1 = text1.lower().split()  # Tokenize the first sentence.
    t2 = text2.lower().split()  # Tokenize the second sentence.
    vocab = np.unique(np.array(t1 + t2))  # Build a shared vocabulary.
    v1 = bag_of_words_vector(t1, vocab)  # Convert the first sentence to counts.
    v2 = bag_of_words_vector(t2, vocab)  # Convert the second sentence to counts.
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))  # Return cosine similarity.
similar_score = cosine_for_pair(sentence_c, sentence_d)  # Compare the similar pair.
different_score = cosine_for_pair(sentence_c, sentence_e)  # Compare the unrelated pair.
print("Similar pair score:", similar_score)  # Display the similar-pair score.
print("Different pair score:", different_score)  # Display the different-pair score.
```

The similar-topic pair should have a higher cosine similarity than the unrelated pair because they share more vocabulary.

```python
# Mini Project: Simple Text Similarity Checker(AP24110011603)
sentence1 = "machine learning uses numpy arrays"  # Store the first sentence.
sentence2 = "machine learning works with numpy arrays"  # Store the second sentence.
sentence3 = "fresh mangoes taste sweet in summer"  # Store the third sentence.

tokens1 = sentence1.lower().split()  # Tokenize sentence 1.(AP24110011603)
tokens2_project = sentence2.lower().split()  # Tokenize sentence 2.
tokens3 = sentence3.lower().split()  # Tokenize sentence 3.

shared_vocab = np.unique(np.array(tokens1 + tokens2_project + tokens3))  # Build one vocabulary for all three sentences.
vector1 = bag_of_words_vector(tokens1, shared_vocab)  # Convert sentence 1 to a bag-of-words vector.
vector2 = bag_of_words_vector(tokens2_project, shared_vocab)  # Convert sentence 2 to a bag-of-words vector.
vector3 = bag_of_words_vector(tokens3, shared_vocab)  # Convert sentence 3 to a bag-of-words vector.

def cosine_similarity(v1, v2):  # Define a cosine-similarity function.
    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)  # Calculate the product of the two vector magnitudes.
    return np.dot(v1, v2) / denominator  # Return the cosine similarity.

score_12 = cosine_similarity(vector1, vector2)  # Compare sentences 1 and 2.
score_13 = cosine_similarity(vector1, vector3)  # Compare sentences 1 and 3.
score_23 = cosine_similarity(vector2, vector3)  # Compare sentences 2 and 3.

scores = {"1-2": score_12, "1-3": score_13, "2-3": score_23}  # Store all three pairwise scores.
best_pair = max(scores, key=scores.get)  # Find the pair with the highest similarity score.

print("Sentence 1:", sentence1)  # Display sentence 1.
print("Sentence 2:", sentence2)  # Display sentence 2.
print("Sentence 3:", sentence3)  # Display sentence 3.
print("Cosine similarity 1-2:", score_12)  # Display the first pair score.
print("Cosine similarity 1-3:", score_13)  # Display the second pair score.
print("Cosine similarity 2-3:", score_23)  # Display the third pair score.
print("Most similar pair:", best_pair)  # Display the pair with the highest score.
```
