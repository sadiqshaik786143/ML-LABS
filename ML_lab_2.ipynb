### Filtering Rows by Column Value

You can filter rows based on conditions applied to specific columns. This is often done using boolean indexing.

```python
# Create a sample DataFrame for demonstration(AP24110011603)
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 28],
        'City': ['New York', 'London', 'Paris', 'New York']}
sample_df = pd.DataFrame(data)

print("Original DataFrame:")
display(sample_df)
```

```python
# Filter rows where 'Age' is greater than 30
filtered_by_age = sample_df[sample_df['Age'] > 30]

print("Filtered by Age > 30:")
display(filtered_by_age)
```

```python
# Filter rows where 'City' is 'New York'
filtered_by_city = sample_df[sample_df['City'] == 'New York']

print("Filtered by City == 'New York':")
display(filtered_by_city)
```

```python
# Filter rows with multiple conditions (Age > 25 and City is 'New York')(AP24110011603)
filtered_by_multiple_conditions = sample_df[(sample_df['Age'] > 25) & (sample_df['City'] == 'New York')]

print("Filtered by Age > 25 AND City == 'New York':")
display(filtered_by_multiple_conditions)
```

So 2:5 selects:

Position 2
Position 3
Position 4

```python
import pandas as pd              # Import Pandas library

import numpy as np               # Import NumPy library

info = pd.DataFrame({             # Create a DataFrame
    "Person": ["Siva", "Pavan", "Gopal", "Raghu"], # Create Person column
    "Age": [27., 29, np.nan, 32] # Create Age column; np.nan represents missing value
})

print(info)                       # Display the DataFrame

print(info.count())               # Count non-null values in each column

print("null values")              # Display a message

print(info.isnull())              # Check every cell for a null value

print(info.isnull().sum())        # Count total null values in each column
```

```python
import pandas as pd              # Import Pandas library

import numpy as np               # Import NumPy library

info = pd.DataFrame({             # Create a DataFrame
    "Person": ["Siva", "Pavan", "Gopal", "Raghu"], # Create Person column
    "Age": [27., 29, np.nan, 32] # Create Age column with one missing value
})

print(info)                       # Display original DataFrame

print("Sorting Order")            # Display sorting message

print(info.sort_index())          # Sort DataFrame by index in ascending order

print("Descending Order")         # Display descending-order message

print(info.sort_index(ascending=False)) # Sort DataFrame by index in descending order
```
